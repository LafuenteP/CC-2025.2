## 2. Em busca de uma solução

Uma **solução** é uma sequência de ações que leva o agente do estado inicial até um estado objetivo, formando uma árvore de busca com estrutura nó. Cada nó contém:

* Estado correspondete

* Pai que gerou o nó

* Ação aplicada para se chegar ao nó

* Custo do caminho para sair do estado inicial e se chegar ao nó

A **borda** é o conjunto de nós disponíveis para expansão. A expansão continua até encontrar a solução ou não haver mais estados a expandir.

## 3. Algoritmos de busca

As estratégias de busca diferem-se em qual estrutura de dados é utilizada para implementar a *borda*

### 3.1 Busca em Largura (BFS)
Implementa a **borda** como uma fila FIFO (*First In First Out*).

```
01. BUSCA-EM-LARGURA(problema){
02.     nó.estado ← estado inicial do problema
03.     nó.custoDeCaminho ← 0
04.     se nó.estado é um estado objetivo então
05.         retorne solução
06.     borda ← fila FIFO contendo nó
07.     explorados ← {}
08.     repita
09.         se borda está vazia então
10.             retorne falha
11.         nó ← remover(borda)
12.         adicionar nó.estado a explorados
13.         para cada ação aplicável
14.             filho ← criarNó(problema, nó, ação)
15.             se filho.estado não está em explorados então
16.                 se filho.estado é objetivo então
17.                     retorne solução
18.                 inserir(filho, borda)
19. }
```


### 3.2 Busca de Custo Uniforme (UCS)
Implementa a **borda** como uma fila de prioridades.

```
01. BUSCA-DE-CUSTO-UNIFORME(problema){
02.     nó.estado ← estado inicial do problema
03.     nó.custoDeCaminho ← 0
04.     borda ← fila de prioridade com nó
05.     explorados ← {}
06.     repita
07.         se borda está vazia então
08.             retorne falha
09.         nó ← remover(borda)
10.         se nó.estado é objetivo então
11.             retorne solução
12.         adicionar nó.estado a explorados
13.         para cada ação aplicável
14.             filho ← criarNó(problema, nó, ação)
15.             se filho.estado não está em borda ou explorados
16.                 inserir(filho, borda)
17.             senão se está na borda com maior custo
18.                 substituir nó por filho
19. }
```

### 3.3 Busca em Profundidade (DFS)

Implementa a **borda** como uma fila LIFO (*Last In First Out*).

```
01. BUSCA-EM-PROFUNDIDADE(problema){
02.     nó.estado ← estado inicial do problema
03.     nó.custoDeCaminho ← 0
04.     se nó.estado é objetivo então
05.         retorne solução
06.     borda ← fila LIFO com nó
07.     explorados ← {}
08.     repita
09.         se borda está vazia então
10.             retorne falha
11.         nó ← remover(borda)
12.         adicionar nó.estado a explorados
13.         para cada ação aplicável
14.             filho ← criarNó(problema, nó, ação)
15.             se filho.estado não está em explorados então
16.                 se filho.estado é objetivo então
17.                     retorne solução
18.                 inserir(filho, borda)
19. }
```

### Exercício 01 - Implemente aqui a busca em largura

````python
import collections

class No:
    def __init__(self, estado: Estado, pai, transicao: Transicao, custo_acumulado: float):
        self.estado = estado
        self.pai = pai
        self.transicao = transicao
        self.custo_acumulado = custo_acumulado

    def __lt__(self, other):
        return self.custo_acumulado < other.custo_acumulado

def reconstruir_caminho(no_final: 'No') -> list:
    caminho = []
    custo_total = no_final.custo_acumulado
    no_atual = no_final
    while no_atual is not None:
        caminho.append(no_atual.estado.Nome)
        no_atual = no_atual.pai
    caminho.reverse()
    return caminho, custo_total


def BuscaEmLargura(Mapa : Mapa, EstadoOrigem : Estado, EstadoFinal : Estado):
  no = No(EstadoOrigem, None, None, 0)
  if(no.estado == EstadoFinal):
    return no.estado
  Borda = collections.deque()
  Explorados = set()
  Borda.append(no)
  while(True):
    if not Borda:
      return False
    no = Borda.popleft()
    Explorados.add(no.estado)
    for transicao in no.estado.Vizinhanca:
      # The cost for BFS is typically considered as the number of steps,
      # not the accumulated edge weights.
      custo_filho = no.custo_acumulado + 1
      filho = No(transicao.Estado, no, transicao, custo_filho)

      # In BFS, we check if the state has been explored to avoid cycles and redundant work.
      # We don't need to check if it's in the border with a higher cost like in UCS.
      if filho.estado not in Explorados:

        if filho.estado == EstadoFinal:

          caminho, custo_total = reconstruir_caminho(filho)
          print("✅ Sucesso!")
          print(f"Caminho encontrado: {caminho}")
          print(f"Número de passos (custo acumulado no BFS): {custo_total}")
          return filho # Retorna o nó objetivo

        Borda.append(filho)

        # Add to explored *after* adding to border to ensure all nodes at a level are explored first
        # However, adding here also works for correctness in this specific problem graph
        # due to finite states and no negative costs. For general graphs, add before appending to border.
        # Keeping it here to match the original logic flow more closely.
        Explorados.add(filho.estado)


if __name__ == '__main__':
    mapa_romenia, estado_inicial, estado_objetivo = main()

    print("\n" + "="*30)
    print(f"BUSCA EM LARGURA de {estado_inicial.Nome} para {estado_objetivo.Nome}")
    print("="*30)

    resultado = BuscaEmLargura(mapa_romenia, estado_inicial, estado_objetivo)

    # You can inspect the result if it's a node, or False
    if resultado is not False:
        # Here, the result already printed the path
        pass
````

### Exercício 02 - Implemente aqui a busca de custo uniforme

````python
import heapq

def BuscaDeCustoUniforme(Mapa: Mapa, EstadoOrigem: Estado, EstadoFinal: Estado):
    no_inicial = No(EstadoOrigem, None, None, 0)
    if no_inicial.estado == EstadoFinal:
        return no_inicial.estado
    Borda = []
    Explorados = {}  # Usando um dicionário para armazenar o custo de cada estado
    heapq.heappush(Borda, no_inicial)
    Explorados[no_inicial.estado] = 0  # O estado inicial com custo 0

    while Borda:
        no = heapq.heappop(Borda)
        if no.estado == EstadoFinal:
            caminho, custo_total = reconstruir_caminho(no)
            print("✅ Sucesso!")
            print(f"Caminho encontrado: {caminho}")
            print(f"Custo total: {custo_total}")
            return no

        for transicao in no.estado.Vizinhanca:
            custo_filho = no.custo_acumulado + transicao.Custo
            filho = No(transicao.Estado, no, transicao, custo_filho)

            # Se o estado filho já foi explorado e encontramos um caminho mais barato
            if filho.estado in Explorados:
                if custo_filho < Explorados[filho.estado]:  # Novo custo é melhor
                    # Atualiza o custo do estado explorado
                    Explorados[filho.estado] = custo_filho
                    # Empurra o filho com o novo custo para a borda
                    heapq.heappush(Borda, filho)
            else:
                # Se o estado filho ainda não foi explorado, adicionamos na borda
                Explorados[filho.estado] = custo_filho
                heapq.heappush(Borda, filho)

    return False

if __name__ == '__main__':
    mapa_romenia, estado_inicial, estado_objetivo = main()

    print("\n" + "="*30)
    print(f"BUSCA DE CUSTO UNIFORME de {estado_inicial.Nome} para {estado_objetivo.Nome}")
    print("="*30)

    resultado = BuscaDeCustoUniforme(mapa_romenia, estado_inicial, estado_objetivo)

    if resultado is not False:
        pass
````
